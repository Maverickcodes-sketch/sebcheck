```python
import os
import torch
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms
from PIL import Image
import matplotlib.pyplot as plt

class PairedImageDataset(Dataset):
    def __init__(self, blur_dir, sharp_dir, target_size=(256, 256)):
        self.blur_dir = blur_dir
        self.sharp_dir = sharp_dir
        self.target_size = target_size

        # Find files present in both directories
        blur_files = set(f for f in os.listdir(blur_dir) if f.lower().endswith(('.png', '.jpg', '.jpeg')))
        sharp_files = set(f for f in os.listdir(sharp_dir) if f.lower().endswith(('.png', '.jpg', '.jpeg')))
        self.matched_filenames = sorted(list(blur_files.intersection(sharp_files)))

        # Filter out any corrupted or unreadable images
        self.valid_filenames = []
        for fname in self.matched_filenames:
            blur_path = os.path.join(self.blur_dir, fname)
            sharp_path = os.path.join(self.sharp_dir, fname)
            try:
                with Image.open(blur_path) as img_b, Image.open(sharp_path) as img_s:
                    img_b.verify()
                    img_s.verify()
                self.valid_filenames.append(fname)
            except (IOError, SyntaxError) as e:
                print(f"Skipping corrupted/unreadable pair for {fname}: {e}")

        print(f"Found and validated {len(self.valid_filenames)} matched image pairs.")

        # Standard transformation pipeline: Resize, Convert to Tensor, and Normalize
        self.transform = transforms.Compose([
            transforms.Resize(self.target_size),
            transforms.ToTensor(),
            transforms.Normalize(mean=[0.5, 0.5, 0.5], std=[0.5, 0.5, 0.5])
        ])

    def __len__(self):
        return len(self.valid_filenames)

    def __getitem__(self, idx):
        fname = self.valid_filenames[idx]
        blur_path = os.path.join(self.blur_dir, fname)
        sharp_path = os.path.join(self.sharp_dir, fname)

        # Open and convert to RGB
        img_blur = Image.open(blur_path).convert('RGB')
        img_sharp = Image.open(sharp_path).convert('RGB')

        # Apply consistent transformations
        tensor_blur = self.transform(img_blur)
        tensor_sharp = self.transform(img_sharp)

        return tensor_blur, tensor_sharp, fname
```

```python
import torch
import torch.nn as nn

class ResidualBlock(nn.Module):
    def __init__(self, channels):
        super(ResidualBlock, self).__init__()
        self.conv1 = nn.Conv2d(channels, channels, kernel_size=3, padding=1, bias=False)
        self.bn1 = nn.BatchNorm2d(channels)
        self.relu = nn.ReLU(inplace=True)
        self.conv2 = nn.Conv2d(channels, channels, kernel_size=3, padding=1, bias=False)
        self.bn2 = nn.BatchNorm2d(channels)

    def forward(self, x):
        residual = x
        out = self.relu(self.bn1(self.conv1(x)))
        out = self.bn2(self.conv2(out))
        out += residual
        return self.relu(out)

class DeblurringGenerator(nn.Module):
    def __init__(self, num_res_blocks=6):
        super(DeblurringGenerator, self).__init__()

        # Initial feature extraction
        self.initial = nn.Sequential(
            nn.Conv2d(3, 64, kernel_size=7, stride=1, padding=3, bias=False),
            nn.BatchNorm2d(64),
            nn.ReLU(inplace=True)
        )

        # Residual blocks to learn mapping
        res_blocks = []
        for _ in range(num_res_blocks):
            res_blocks.append(ResidualBlock(64))
        self.res_blocks = nn.Sequential(*res_blocks)

        # Final reconstruction to output an RGB image in range [-1, 1]
        self.final = nn.Sequential(
            nn.Conv2d(64, 3, kernel_size=7, stride=1, padding=3),
            nn.Tanh()
        )

    def forward(self, x):
        out = self.initial(x)
        out = self.res_blocks(out)
        out = self.final(out)
        return out
```

```python
# Instantiate the generator and perform a quick forward pass test
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
generator = DeblurringGenerator(num_res_blocks=6).to(device)

# Put generator in evaluation mode for a quick validation pass
generator.eval()

# Take the blurry batch from our previously loaded DataLoader and move to device
blur_inputs = blur_batch.to(device)

with torch.no_grad():
    restored_outputs = generator(blur_inputs)

print(f"Device: {device}")
print(f"Input shape (blur): {blur_inputs.shape}")
print(f"Output shape (restored): {restored_outputs.shape}")
print(f"Output pixel range: Min={restored_outputs.min().item():.4f}, Max={restored_outputs.max().item():.4f}")
```

```python
import torch
import torch.nn as nn

class PatchGANDiscriminator(nn.Module):
    def __init__(self, in_channels=6, ndf=64):
        super(PatchGANDiscriminator, self).__init__()

        # Standard PatchGAN structure: successive convolutional blocks downsampling spatial dimensions
        self.model = nn.Sequential(
            # Input shape: [batch, 6, H, W] -> outputs: [batch, 64, H/2, W/2]
            nn.Conv2d(in_channels, ndf, kernel_size=4, stride=2, padding=1),
            nn.LeakyReLU(0.2, inplace=True),

            # [batch, 64, H/2, W/2] -> [batch, 128, H/4, W/4]
            nn.Conv2d(ndf, ndf * 2, kernel_size=4, stride=2, padding=1, bias=False),
            nn.BatchNorm2d(ndf * 2),
            nn.LeakyReLU(0.2, inplace=True),

            # [batch, 128, H/4, W/4] -> [batch, 256, H/8, W/8]
            nn.Conv2d(ndf * 2, ndf * 4, kernel_size=4, stride=2, padding=1, bias=False),
            nn.BatchNorm2d(ndf * 4),
            nn.LeakyReLU(0.2, inplace=True),

            # [batch, 256, H/8, W/8] -> [batch, 512, H/8, W/8] (stride=1 retains spatial dimension here)
            nn.Conv2d(ndf * 4, ndf * 8, kernel_size=4, stride=1, padding=1, bias=False),
            nn.BatchNorm2d(ndf * 8),
            nn.LeakyReLU(0.2, inplace=True),

            # Final 1-channel classification grid output
            nn.Conv2d(ndf * 8, 1, kernel_size=4, stride=1, padding=1)
        )

    def forward(self, blurry_img, candidate_img):
        # Concatenate along the channel dimension (dim=1) to construct conditional pairs
        x = torch.cat([blurry_img, candidate_img], dim=1)
        return self.model(x)
```

```python
# Instantiate the PatchGAN discriminator
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
discriminator = PatchGANDiscriminator().to(device)
discriminator.eval()

# Take the real sharp batch to verify discriminator inputs
blur_inputs = blur_batch.to(device)
sharp_inputs = sharp_batch.to(device)

# Ensure generator is loaded and evaluated
generator = generator.to(device)
generator.eval()
with torch.no_grad():
    restored_outputs = generator(blur_inputs)
    # 1. Forward pass on real pair: (I_blurry, I_sharp)
    real_prediction = discriminator(blur_inputs, sharp_inputs)
    # 2. Forward pass on generated fake pair: (I_blurry, G(I_blurry))
    fake_prediction = discriminator(blur_inputs, restored_outputs)

print(f"Device: {device}")
print(f"Discriminator inputs: concatenated shape {torch.cat([blur_inputs, sharp_inputs], dim=1).shape}")
print(f"Real pair patch score grid shape: {real_prediction.shape}")
print(f"Fake pair patch score grid shape: {fake_prediction.shape}")
```

```python
import torch
import torch.nn as nn

class LossManager:
    def __init__(self, lambda_L1=100.0, use_lsgan=True):
        """
        Manages losses for training a conditional GAN for image restoration.

        Args:
            lambda_L1 (float): Weight factor for pixel-level L1 reconstruction loss.
            use_lsgan (bool): If True, uses Mean Squared Error (LSGAN) loss.
                              If False, uses standard Binary Cross Entropy (BCE) loss.
        """
        self.lambda_L1 = lambda_L1
        self.use_lsgan = use_lsgan

        # Define adversarial loss objective
        if self.use_lsgan:
            self.adversarial_loss = nn.MSELoss()
            print("Using Least-Squares GAN (LSGAN) objective.")
        else:
            self.adversarial_loss = nn.BCEWithLogitsLoss()
            print("Using Binary Cross Entropy (BCE) objective.")

        # Define reconstruction loss objective
        self.l1_loss = nn.L1Loss()

    def get_discriminator_loss(self, real_preds, fake_prediction):
        """
        Calculates the discriminator's loss objective:
        L_D = L_real + L_fake
        """
        # Real predictions should aim for a target of 1.0
        real_targets = torch.ones_like(real_preds)
        loss_real = self.adversarial_loss(real_preds, real_targets)

        # Fake predictions should aim for a target of 0.0
        fake_targets = torch.zeros_like(fake_prediction)
        loss_fake = self.adversarial_loss(fake_prediction, fake_targets)

        total_d_loss = loss_real + loss_fake
        return total_d_loss, loss_real, loss_fake

    def get_generator_loss(self, fake_prediction, generated_img, real_sharp_img):
        """
        Calculates the generator's multi-objective loss:
        L_G = L_adversarial + lambda_L1 * L_L1
        """
        # Adversarial loss: Generator wants fake predictions to be classified as real (1.0)
        real_targets = torch.ones_like(fake_prediction)
        loss_adv = self.adversarial_loss(fake_prediction, real_targets)

        # Reconstruction loss: Pixel-to-pixel absolute difference with ground truth
        loss_rec = self.l1_loss(generated_img, real_sharp_img)

        # Weighted sum of losses
        total_g_loss = loss_adv + (self.lambda_L1 * loss_rec)
        return total_g_loss, loss_adv, loss_rec
```

```python
# Instantiate and dry-run test the combined losses on our loaded batch sample
loss_manager = LossManager(lambda_L1=100.0, use_lsgan=True)

# Calculate dry-run losses on current batch outputs
d_loss, d_real, d_fake = loss_manager.get_discriminator_loss(real_prediction, fake_prediction)
g_loss, g_adv, g_rec = loss_manager.get_generator_loss(fake_prediction, restored_outputs, sharp_inputs)

print(f"\n--- Loss Verification Dry-Run Results ---")
print(f"Discriminator Loss: {d_loss.item():.4f} (Real component: {d_real.item():.4f}, Fake component: {d_fake.item():.4f})")
print(f"Generator Loss: {g_loss.item():.4f} (Adversarial: {g_adv.item():.4f}, L1 Reconstruction: {g_rec.item():.4f})")
```

```python
import torch
import torch.optim as optim
from tqdm.auto import tqdm
import matplotlib.pyplot as plt

# --- Training Configurations ---
EPOCHS =10      # Start with a small number to validate the pipeline
LR_G = 2e-4            # Learning rate for Generator
LR_D = 2e-4            # Learning rate for Discriminator
BETA1 = 0.5            # Momentum term for Adam optimizer
BETA2 = 0.999
VIS_INTERVAL = 1       # Visualize results at the end of each epoch

# Ensure models are on the correct device
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
generator = generator.to(device)
discriminator = discriminator.to(device)

# Separate Optimizers
optimizer_G = optim.Adam(generator.parameters(), lr=LR_G, betas=(BETA1, BETA2))
optimizer_D = optim.Adam(discriminator.parameters(), lr=LR_D, betas=(BETA1, BETA2))

# Containers to track loss history
train_losses_G = []
train_losses_D = []
val_losses_G = []
val_losses_D = []

def visualize_epoch_results(epoch, blur_batch, sharp_batch, generator, device):
    """Helper function to visualize progress on a fixed batch at the end of an epoch."""
    generator.eval()
    with torch.no_grad():
        blur_inputs = blur_batch.to(device)
        restored_outputs = generator(blur_inputs).cpu()

    fig, axes = plt.subplots(1, 3, figsize=(15, 5))
    # Denormalize [-1, 1] to [0, 1] for rendering
    img_blur = blur_batch[0].permute(1, 2, 0).numpy() * 0.5 + 0.5
    img_restored = restored_outputs[0].permute(1, 2, 0).numpy() * 0.5 + 0.5
    img_sharp = sharp_batch[0].permute(1, 2, 0).numpy() * 0.5 + 0.5

    axes[0].imshow(img_blur)
    axes[0].set_title("Blurry Input")
    axes[0].axis('off')

    axes[1].imshow(img_restored)
    axes[1].set_title(f"Restored (Epoch {epoch+1})")
    axes[1].axis('off')

    axes[2].imshow(img_sharp)
    axes[2].set_title("Ground Truth")
    axes[2].axis('off')

    plt.tight_layout()
    plt.show()

# --- Complete Training Loop with Error Handling ---
for epoch in range(EPOCHS):
    generator.train()
    discriminator.train()

    running_g_loss = 0.0
    running_d_loss = 0.0

    # Use tqdm progress bar for batch loops
    progress_bar = tqdm(train_loader, desc=f"Epoch {epoch+1}/{EPOCHS}")

    try:
        for batch_idx, (blur_imgs, sharp_imgs, _) in enumerate(progress_bar):
            blur_imgs = blur_imgs.to(device)
            sharp_imgs = sharp_imgs.to(device)

            batch_curr_size = blur_imgs.size(0)

            # ==========================================
            # 1. Train Discriminator: L_D = L_real + L_fake
            # ==========================================
            optimizer_D.zero_grad()

            # Real pair forward pass
            real_preds = discriminator(blur_imgs, sharp_imgs)

            # Generate restored images
            fake_imgs = generator(blur_imgs)

            # Fake pair forward pass (detach to avoid backpropping through generator here)
            fake_preds = discriminator(blur_imgs, fake_imgs.detach())

            # Loss calculation
            d_loss, _, _ = loss_manager.get_discriminator_loss(real_preds, fake_preds)
            d_loss.backward()
            optimizer_D.step()

            # ==========================================
            # 2. Train Generator: L_G = L_adv + lambda * L_L1
            # ==========================================
            optimizer_G.zero_grad()

            # Discriminator predictions on fake images without detachment
            fake_preds_for_g = discriminator(blur_imgs, fake_imgs)

            # Loss calculation
            g_loss, _, _ = loss_manager.get_generator_loss(fake_preds_for_g, fake_imgs, sharp_imgs)
            g_loss.backward()
            optimizer_G.step()

            # Accumulate metrics
            running_g_loss += g_loss.item() * batch_curr_size
            running_d_loss += d_loss.item() * batch_curr_size

            # Update progress bar suffix
            progress_bar.set_postfix({
                'G_loss': f"{g_loss.item():.4f}",
                'D_loss': f"{d_loss.item():.4f}"
            })

    except RuntimeError as e:
        if "out of memory" in str(e):
            print("WARNING: Out of memory during training batch. Clearing cache.")
            if torch.cuda.is_available():
                torch.cuda.empty_cache()
            continue
        else:
            raise e

    # Calculate epoch average training loss
    epoch_g_loss = running_g_loss / len(train_set)
    epoch_d_loss = running_d_loss / len(train_set)
    train_losses_G.append(epoch_g_loss)
    train_losses_D.append(epoch_d_loss)

    # --- Validation Loop ---
    generator.eval()
    discriminator.eval()
    val_g_loss = 0.0
    val_d_loss = 0.0

    with torch.no_grad():
        for val_blur, val_sharp, _ in val_loader:
            val_blur = val_blur.to(device)
            val_sharp = val_sharp.to(device)
            val_batch_size = val_blur.size(0)

            val_fake = generator(val_blur)

            real_preds_val = discriminator(val_blur, val_sharp)
            fake_preds_val = discriminator(val_blur, val_fake)

            d_loss_val, _, _ = loss_manager.get_discriminator_loss(real_preds_val, fake_preds_val)
            g_loss_val, _, _ = loss_manager.get_generator_loss(fake_preds_val, val_fake, val_sharp)

            val_g_loss += g_loss_val.item() * val_batch_size
            val_d_loss += d_loss_val.item() * val_batch_size

    epoch_val_g = val_g_loss / len(val_set)
    epoch_val_d = val_d_loss / len(val_set)
    val_losses_G.append(epoch_val_g)
    val_losses_D.append(epoch_val_d)

    print(f"Epoch [{epoch+1}/{EPOCHS}] Summary:")
    print(f"  Train -> G Loss: {epoch_g_loss:.4f} | D Loss: {epoch_d_loss:.4f}")
    print(f"  Val   -> G Loss: {epoch_val_g:.4f} | D Loss: {epoch_val_d:.4f}")

    # Visual verification at intervals
    if (epoch + 1) % VIS_INTERVAL == 0:
        visualize_epoch_results(epoch, blur_batch, sharp_batch, generator, device)
```

```python
import numpy as np
from skimage.metrics import peak_signal_noise_ratio as psnr_metric
from skimage.metrics import structural_similarity as ssim_metric
import torch
from tqdm.auto import tqdm

def evaluate_metrics(generator, dataloader, device):
    generator.eval()

    blurry_psnrs = []
    blurry_ssims = []
    restored_psnrs = []
    restored_ssims = []

    with torch.no_grad():
        for blur_imgs, sharp_imgs, _ in tqdm(dataloader, desc="Evaluating"):
            blur_imgs = blur_imgs.to(device)
            restored_imgs = generator(blur_imgs).cpu()

            # Convert tensors to numpy array in range [0, 1] with shape [H, W, C]
            # Tensors are normalized with mean=0.5, std=0.5
            blur_np = (blur_imgs.cpu().numpy() * 0.5 + 0.5).transpose(0, 2, 3, 1)
            sharp_np = (sharp_imgs.numpy() * 0.5 + 0.5).transpose(0, 2, 3, 1)
            restored_np = (restored_imgs.numpy() * 0.5 + 0.5).transpose(0, 2, 3, 1)

            for i in range(blur_np.shape[0]):
                b_img = np.clip(blur_np[i], 0.0, 1.0)
                s_img = np.clip(sharp_np[i], 0.0, 1.0)
                r_img = np.clip(restored_np[i], 0.0, 1.0)

                # Specify channel_axis for color images in SSIM
                blurry_psnrs.append(psnr_metric(s_img, b_img, data_range=1.0))
                blurry_ssims.append(ssim_metric(s_img, b_img, data_range=1.0, channel_axis=2))

                restored_psnrs.append(psnr_metric(s_img, r_img, data_range=1.0))
                restored_ssims.append(ssim_metric(s_img, r_img, data_range=1.0, channel_axis=2))

    return {
        'blurry_psnr_mean': np.mean(blurry_psnrs), 'blurry_psnr_std': np.std(blurry_psnrs),
        'blurry_ssim_mean': np.mean(blurry_ssims), 'blurry_ssim_std': np.std(blurry_ssims),
        'restored_psnr_mean': np.mean(restored_psnrs), 'restored_psnr_std': np.std(restored_psnrs),
        'restored_ssim_mean': np.mean(restored_ssims), 'restored_ssim_std': np.std(restored_ssims)
    }

metrics = evaluate_metrics(generator, test_loader, device)

print("=== Quantitative Evaluation (Test Set) ===")
print(f"Original Blurry vs Sharp Reference:")
print(f"  PSNR: {metrics['blurry_psnr_mean']:.3f} ± {metrics['blurry_psnr_std']:.3f} dB")
print(f"  SSIM: {metrics['blurry_ssim_mean']:.4f} ± {metrics['blurry_ssim_std']:.4f}")
print(f"GAN-Restored vs Sharp Reference:")
print(f"  PSNR: {metrics['restored_psnr_mean']:.3f} ± {metrics['restored_psnr_std']:.3f} dB")
print(f"  SSIM: {metrics['restored_ssim_mean']:.4f} ± {metrics['restored_ssim_std']:.4f}")
```

### Limitations of Pixel-Level Metrics for License Plate Restoration

While **PSNR** (Peak Signal-to-Noise Ratio) and **SSIM** (Structural Similarity Index) are standard benchmarks in image deblurring, they possess significant limitations when assessing the success of license plate restoration:

1. **Sensitivity to Pixel Alignment over Semantic Integrity**: PSNR is based directly on Mean Squared Error (MSE), which penalizes pixel-level spatial shifts. A restored image that is slightly offset spatially might yield a poor PSNR score despite containing sharp, highly readable text characters.
2. **SSIM Structural Bias**: SSIM evaluates luminance, contrast, and structure. While it correlates better with human visual perception than MSE, it still heavily favors overall structural agreement over the micro-features necessary to distinguish ambiguous alphanumeric characters (e.g., telling a 'B' from an '8', or an 'O' from a 'D').
3. **The "Hallucination" Paradox**: Conditional GANs are optimized to generate plausible, sharp, and high-frequency textures. However, high perceptual quality or a visually "sharper" plate does not guarantee accuracy of the underlying content. GANs can easily hallucinate incorrect stroke details to construct realistic-looking but entirely fabricated alphanumeric characters, which may artificially inflate metrics without restoring the authentic plate information.

```python
import torch
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt
from torchvision import transforms

def restore_blurry_plate(image_path, generator, device, target_size=(128, 256)):
    """
    Restores a single blurry number plate image using the trained generator.

    Args:
        image_path (str): Path to the blurry image file.
        generator (nn.Module): The trained generator model.
        device (torch.device): Device to run inference on (CPU or GPU).
        target_size (tuple): Preprocessing image target size (H, W).
    """
    # 1. Load and preprocess the image
    img_raw = Image.open(image_path).convert('RGB')

    # Apply the same transformations as used during training
    preprocess = transforms.Compose([
        transforms.Resize(target_size),
        transforms.ToTensor(),
        transforms.Normalize(mean=[0.5, 0.5, 0.5], std=[0.5, 0.5, 0.5])
    ])

    input_tensor = preprocess(img_raw).unsqueeze(0).to(device)  # Add batch dimension

    # 2. Run inference in evaluation mode without tracking gradients
    generator.eval()
    with torch.no_grad():
        output_tensor = generator(input_tensor).cpu().squeeze(0)  # Remove batch dimension

    # 3. Postprocess and convert back to numpy image range [0, 1]
    input_np = (input_tensor.cpu().squeeze(0).numpy() * 0.5 + 0.5).transpose(1, 2, 0)
    output_np = (output_tensor.numpy() * 0.5 + 0.5).transpose(1, 2, 0)

    input_np = np.clip(input_np, 0.0, 1.0)
    output_np = np.clip(output_np, 0.0, 1.0)

    # 4. Display the input blurry and GAN-restored images side-by-side
    fig, axes = plt.subplots(1, 2, figsize=(12, 6))

    axes[0].imshow(input_np)
    axes[0].set_title("Original Blurry Input")
    axes[0].axis('off')

    axes[1].imshow(output_np)
    axes[1].set_title("GAN-Restored Output")
    axes[1].axis('off')

    plt.tight_layout()
    plt.show()

# Quick test using a sample blurry image path
sample_blur_path = os.path.join(blur_dir, 'grab1635.jpg')
restore_blurry_plate(sample_blur_path, generator, device)
# Quick test using a sample blurry image path
sample_blur_path = os.path.join(blur_dir, 'grab1637.jpg')
restore_blurry_plate(sample_blur_path, generator, device)
# Quick test using a sample blurry image path
sample_blur_path = os.path.join(blur_dir, 'grab1639.jpg')
restore_blurry_plate(sample_blur_path, generator, device)
```
